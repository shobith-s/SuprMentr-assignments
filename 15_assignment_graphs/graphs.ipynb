```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
import pandas as pd

import warnings
warnings.filterwarnings("ignore")

# Load the dataset
data = load_iris()
df = pd.DataFrame(data.data, columns=data.feature_names)
df['species'] = data.target
df['species_name'] = df['species'].map({0: 'Setosa', 1: 'Versicolor', 2: 'Virginica'})

# Set the visual style
sns.set_theme(style="whitegrid")
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(x='species_name', y='petal length (cm)', data=df, palette='viridis', ci=None)
plt.title('Average Petal Length Across Iris Species', fontsize=14)
plt.xlabel('Species', fontsize=12)
plt.ylabel('Petal Length (cm)', fontsize=12)
plt.show()
```

```python
species_counts = df['species_name'].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(species_counts, labels=species_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
plt.title('Distribution of Samples in the Study')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['sepal width (cm)'], kde=True, color='teal', bins=15)
plt.title('The "Normalcy" of Sepal Widths', fontsize=14)
plt.xlabel('Sepal Width (cm)', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.show()
```

- **The Identification (Bar Chart):** The bar chart tells us that Petal Length is the most reliable "ID card" for these species. The Setosa is tiny and compact, while the Virginica is the giant of the group. If you find an Iris with a petal longer than 5cm, you are almost certainly looking at a Virginica.

- **The Reliability (Pie Chart):** Our pie chart confirms that our conclusions aren't skewed toward one specific type of flower. Because each species represents exactly 33.3% of the data, our comparisons are "fair" and scientifically sound.

- **The Consistency (Histogram):** Finally, the histogram shows us a "Normal Distribution" (the classic bell curve) for sepal widths. Most flowers hover around the 3.0 cm mark. This tells us that while petal length varies wildly between species, sepal width is a more "stable" trait across the entire Iris family.
