```python
import pandas as pd
import numpy as np
from sklearn import datasets
from sklearn.datasets import load_iris
```

```python
data = load_iris()
```

```python
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target
```

```python
# Display top 2 rows of the dataframe
df.head(2)
```

```python
print(df.max())
print(df['petal length (cm)'].max())
print(df.max(numeric_only=True).max())
```

```python
null_counts = df.isnull().sum()
print(null_counts)
```

```python
class_counts = df["target"].value_counts().sort_index()
petal_means = df.groupby("target")["petal length (cm)"].mean()
sepal_diff = df.groupby("target")["sepal length (cm)"].mean().loc[2] - df.groupby("target")["sepal length (cm)"].mean().loc[0]
petal_diff = petal_means.loc[2] - petal_means.loc[0]

print(f"1. The Iris dataset contains {df.shape[0]} samples with {len(data.feature_names)} numerical features and 1 target label.")
print(f"2. Data quality is strong: total missing values across all columns = {int(null_counts.sum())}.")
print(f"3. The target classes are perfectly balanced: {class_counts.to_dict()} (50 samples per class).")
print(f"4. Mean petal length increases consistently by class: setosa={petal_means.loc[0]:.2f}, versicolor={petal_means.loc[1]:.2f}, virginica={petal_means.loc[2]:.2f}.")
print(f"5. Petal length separates classes more strongly than sepal length (class 2 vs 0 difference: petal={petal_diff:.2f} cm, sepal={sepal_diff:.2f} cm).")
```
