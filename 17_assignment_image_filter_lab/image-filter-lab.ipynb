# Image Filter Lab
Using OpenCV to apply common image processing filters: Grayscale, Blur, and Edge Detection.

```python
import cv2
import matplotlib.pyplot as plt

# Load image
image = cv2.imread('sample.png')
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# 1. Grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# 2. Blur (Gaussian)
blur = cv2.GaussianBlur(image_rgb, (15, 15), 0)

# 3. Edge Detection (Canny)
edges = cv2.Canny(gray, 100, 200)

# Show comparison
plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.imshow(image_rgb)
plt.title('Original')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(gray, cmap='gray')
plt.title('Grayscale')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(blur)
plt.title('Gaussian Blur')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(edges, cmap='gray')
plt.title('Canny Edge Detection')
plt.axis('off')

plt.tight_layout()
plt.show()
```

## Summary of Filters

1. **Grayscale**: Reduces complexity from 3 channels to 1, focusing on intensity. Often a preprocessing step for AI.
2. **Blur**: Smooths out pixel noise and detail. Useful for background removal or anonymization.
3. **Edge Detection**: Highlights the boundaries of objects by detecting sharp changes in brightness. Foundational for object recognition.
