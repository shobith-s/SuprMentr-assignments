```python
import re
import string
import nltk
from nltk.corpus import stopwords

# Download the necessary stopword list
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))
```

```python
def text_cleaner(text):
    # 1. Lowercase the text
    text = text.lower()
    
    # 2. Remove punctuation using regex
    # This replaces any punctuation character with an empty string
    text = text.translate(str.maketrans('', '', string.punctuation))
    
    # 3. Tokenize and remove stopwords
    words = text.split()
    cleaned_words = [w for w in words if w not in stop_words]
    
    # Join back into a string or return as a list
    return " ".join(cleaned_words)
```

```python
# --- Testing the Cleaner ---
test_sentences = [
    "Hello! This is a test sentence, showing how the cleaner works.",
    "Data Science is AMAZING, but cleaning data is 90% of the work!",
    "Can you believe it? Python makes NLP so much easier."
]

print(f"{'RAW TEXT':<60} | {'CLEANED OUTPUT'}")
print("-" * 100)
for s in test_sentences:
    print(f"{s:<60} | {text_cleaner(s)}")
```
