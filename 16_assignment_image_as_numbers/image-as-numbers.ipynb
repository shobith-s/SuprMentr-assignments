# Image as Numbers
Computer sees images as matrices of numbers. This notebook explores an image's shape, pixel values, and color channels.

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

# Load the image
image_path = 'sample.png'
image = cv2.imread(image_path)

if image is None:
    print("Error: Image not found!")
else:
    # OpenCV loads images in BGR format, convert to RGB for plotting
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    
    # 1. Print Shape
    print(f"Image Shape: {image.shape}")
    print(f"Height: {image.shape[0]}, Width: {image.shape[1]}, Channels: {image.shape[2]}")
    
    # 2. Print Pixel Values (small slice)
    print("\nPixel Values for a small 3x3 region (Top-Left):")
    print(image_rgb[0:3, 0:3])
    
    # Show the image
    plt.imshow(image_rgb)
    plt.title('Original Image')
    plt.axis('off')
    plt.show()
```

## Explanation

1. **Shape**: An image is a 3D array (Height, Width, Color Channels). 
2. **Pixel Values**: Each pixel is represented by numbers, typically ranging from 0 (black) to 255 (full intensity).
3. **Channels**: A standard color image has 3 channels: **Red, Green, and Blue (RGB)**. Each channel is a 2D layer of intensity values. For example, a pure red pixel would be `(255, 0, 0)`.
