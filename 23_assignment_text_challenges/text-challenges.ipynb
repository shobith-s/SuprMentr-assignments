```python
import re

messy_data = [
    "OMG this is so cool!! 😍🔥",
    "wht r u doin tday?",
    "The product quality is rly bad... smh.",
    "I luv it!!!!!!!!! 💖",
    "thx for the help, u r the best.",
    "i'm so sry i forgot to call u.",
    "This is lit af! 💯",
    "can't wait for d weekend!!",
    "Y r u so l8?",
    "brb, gotta go to d store.",
    "idk what u mean lol.",
    "This is a gr8 deal! 💸",
    "plz send me the file asap.",
    "no prob, i got u.",
    "That movie was boring af tbh.",
    "hbd!! hope u have a gr8 day! 🎂",
    "stfu i dont believe it!! 😲",
    "i'm omw right now.",
    "wanna go out later?",
    "this is def the best pizza evr. 🍕"
]

slang_dict = {
    "u": "you", "r": "are", "wht": "what", "doin": "doing", "tday": "today",
    "rly": "really", "smh": "shaking my head", "luv": "love", "thx": "thanks",
    "sry": "sorry", "af": "as f***", "d": "the", "y": "why", "l8": "late",
    "brb": "be right back", "idk": "i do not know", "lol": "laughing out loud",
    "gr8": "great", "plz": "please", "asap": "as soon as possible",
    "prob": "problem", "tbh": "to be honest", "hbd": "happy birthday",
    "stfu": "shut the f*** up", "omw": "on my way", "wanna": "want to",
    "def": "definitely", "evr": "ever"
}
```

```python
def preprocess_text(text):
    # a. Lowercasing
    text = text.lower()
    
    # b. Remove Emojis/Non-ASCII (Alternative: Emoji-to-text conversion)
    text = text.encode('ascii', 'ignore').decode('ascii')
    
    # c. Remove excessive punctuation
    text = re.sub(r'([!?.])\1+', r'\1', text)
    
    # d. Slang/Typo Correction
    words = text.split()
    corrected_words = [slang_dict.get(w, w) for w in words]
    
    return " ".join(corrected_words).strip()

# 3. Execution & Comparison
print(f"{'ORIGINAL SENTENCE':<35} | {'CLEANED SENTENCE'}")
print("-" * 75)
for sentence in messy_data:
    cleaned = preprocess_text(sentence)
    print(f"{sentence:<35} | {cleaned}")
```
