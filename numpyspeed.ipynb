```python
import numpy as np
import time

# Python list
list_data = list(range(1_000_000))
start_time = time.time()
list_result = [x*2 for x in list_data]   # simple operation
end_time = time.time()
print("Python list time:", end_time - start_time)

# NumPy array
array_data = np.arange(1_000_000)
start_time = time.time()
array_result = array_data * 2            # vectorized operation
end_time = time.time()
print("NumPy array time:", end_time - start_time)
```

- **Speed:** NumPy array time is noticeably lower than the Python list time — vectorized C-backed operations remove Python-loop overhead.  
- **Scalability:** The gap widens with large sizes (1,000,000 items here), so NumPy scales far better for bulk numeric work.  
- **Memory & overhead:** Python lists create many Python objects (higher memory and per-item overhead); NumPy's contiguous numeric arrays use less overhead, enabling faster, more cache-friendly computation.
