# Movie Review Sentiment Analyzer
This notebook builds a simple rule-based sentiment analyzer and tests it on 5 sample movie reviews.

```python
def simple_sentiment_analyzer(text):
    """A simple rule-based sentiment analyzer."""
    positive_words = {'excellent', 'amazing', 'good', 'great', 'brilliant', 'wonderful', 'best', 'love', 'perfect'}
    negative_words = {'bad', 'terrible', 'worst', 'poor', 'awful', 'boring', 'hated', 'dull', 'waste'}
    
    words = text.lower().split()
    pos_count = sum(1 for word in words if word in positive_words)
    neg_count = sum(1 for word in words if word in negative_words)
    
    if pos_count > neg_count:
        return "Positive"
    elif neg_count > pos_count:
        return "Negative"
    else:
        return "Neutral"

# 5 sample reviews
reviews = [
    "The movie was excellent and the acting was amazing!",
    "I hated this film, it was a total waste of time and boring.",
    "The plot was good, but the ending was a bit dull.",
    "A brilliant masterpiece that I would recommend to everyone!",
    "The cinematography was poor and the story was terrible."
]

print("Movie Review Sentiment Analysis Results:\n")
for review in reviews:
    sentiment = simple_sentiment_analyzer(review)
    print(f"Review: {review}")
    print(f"Sentiment: {sentiment}")
    print("-" * 20)
```

## Conclusion
The simple rule-based system works well for clear opinions but might struggle with sarcasm or complex sentence structures. For 5 reviews, it correctly identified the general tone based on key adjectives.
