```python
from sklearn.datasets import fetch_california_housing
import pandas as pd

housing = fetch_california_housing()

df = pd.DataFrame(housing.data, columns=housing.feature_names)
df['MedHouseVal'] = housing.target

print(df.head())

import warnings
warnings.filterwarnings("ignore")
```

```python
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

X = df[['MedInc']]
y = df['MedHouseVal']

model = LinearRegression()
model.fit(X, y)

plt.figure(figsize=(8, 5))
plt.scatter(X, y, alpha=0.3, color='teal', s=10)
plt.plot(X, model.predict(X), color='orange', linewidth=3)
plt.title("Income vs. House Value (California)")
plt.xlabel("Median Income (Tens of Thousands)")
plt.ylabel("House Value ($100k units)")
plt.show()

test_income = [[5.0]]
prediction = model.predict(test_income)
print(f"Predicted House Value: ${prediction[0]*100000:,.2f}")
```

Why data cleaning matters ?

- Accuracy: Duplicates (like Customer 101 appearing twice) artificially inflate your sales numbers, leading to false "growth" reports.

- Model Stability: Most ML algorithms (like the Linear Regression we built) will literally crash if they encounter a None or NaN value.

- Searchability: If one entry says "Great!" and another says "great", a computer treats them as two different emotions. Standardizing text allows us to group them into a single, powerful "Positive" category.
