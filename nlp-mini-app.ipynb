# NLP Mini App: Keyword Extractor
This app extracts the most relevant keywords from a block of text using a frequency-based approach.

```python
import re
from collections import Counter

def extract_keywords(text, top_n=5):
    # Remove special characters and numbers
    text = re.sub(r'[^a-zA-Z\s]', '', text).lower()
    
    # Basic stop words list
    stop_words = {'the', 'a', 'an', 'and', 'but', 'if', 'or', 'because', 'as', 'what', 
                  'which', 'this', 'that', 'with', 'for', 'from', 'in', 'on', 'to', 'is', 'was'}
    
    words = text.split()
    # Filter out stop words and short words
    filtered_words = [word for word in words if word not in stop_words and len(word) > 3]
    
    # Count frequencies
    word_counts = Counter(filtered_words)
    return word_counts.most_common(top_n)

# Sample input text
sample_text = """
Artificial intelligence is a branch of computer science that aims to create intelligent machines. 
It has become an essential part of the technology industry. 
Machine learning and deep learning are patterns of artificial intelligence that are used to build such systems.
These systems can process vast amounts of data to provide insights and automation.
"""

print("Extracting Keywords...\n")
keywords = extract_keywords(sample_text)

for i, (word, freq) in enumerate(keywords):
    print(f"{i+1}. {word.capitalize()} (Frequency: {freq})")
```

## Usefulness
Keyword extractors are used in SEO optimization, document summarization, and tag generation for blog posts or academic papers.
