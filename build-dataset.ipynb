```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

import warnings
warnings.filterwarnings("ignore")
```

```python
np.random.seed(42)
study_hours = np.random.uniform(1, 10, 50)
marks = 10 + (8 * study_hours) + np.random.normal(0, 5, 50)
```

```python
df = pd.DataFrame({'Study_Hours': study_hours, 'Marks': marks})
print(df.head())
```

- **Feature (X)**: The independent variable or "input." Here, it is Study_Hours. This is what we use to make the prediction.

- **Label (y)**: The dependent variable or "output." Here, it is Marks. This is the value we are trying to predict.

```python
X = df[['Study_Hours']]
y = df['Marks']

model = LinearRegression()
model.fit(X, y)

new_student = np.array([[7.5]])
predicted_mark = model.predict(new_student)

print(f"Prediction for 7.5 hours of study: {predicted_mark[0]:.2f}%")
```
