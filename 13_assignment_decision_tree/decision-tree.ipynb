```python
from sklearn.tree import DecisionTreeClassifier, plot_tree
import pandas as pd
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings("ignore")
```

```python
data = {
    'Is_Raining': [0, 1, 0, 0, 1, 0, 0, 1],
    'Homework_Done': [1, 1, 0, 1, 0, 1, 1, 0],
    'Is_Cold': [0, 0, 0, 1, 1, 0, 1, 0],
    'Play_Outside': [1, 0, 0, 0, 0, 1, 0, 0] # 1 = Play, 0 = Stay
}

df = pd.DataFrame(data)
X = df[['Is_Raining', 'Homework_Done', 'Is_Cold']]
y = df['Play_Outside']
```

```python
clf = DecisionTreeClassifier(criterion='entropy', random_state=42)
clf.fit(X, y)

plt.figure(figsize=(10, 8))
plot_tree(clf, feature_names=X.columns, class_names=['Stay', 'Play'], filled=True, rounded=True)
plt.title("Decision Tree: Should I Play Outside?")
plt.show()
```

```python
prediction = clf.predict([[0, 1, 0]])
result = "PLAY OUTSIDE!" if prediction[0] == 1 else "STAY INSIDE."
print(f"The machine says: {result}")
```
