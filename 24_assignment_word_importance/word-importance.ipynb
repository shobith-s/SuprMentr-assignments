# Word Importance Explorer
This notebook uses TF-IDF (Term Frequency-Inverse Document Frequency) to identify the most important keywords across 5 sample documents.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

# 5 sample documents
documents = [
    "The quick brown fox jumps over the lazy dog.",
    "Artificial Intelligence is transforming the world of technology and automation.",
    "Machine learning models require high-quality data to produce accurate predictions.",
    "Space exploration has seen significant milestones with private companies launching rockets.",
    "The ecosystem depends on the balance between various climate factors and biodiversity."
]

# Initialize TF-IDF Vectorizer
vectorizer = TfidfVectorizer(stop_words='english')

# Fit and transform the documents
tfidf_matrix = vectorizer.fit_transform(documents)

# Get feature names (words)
feature_names = vectorizer.get_feature_names_out()

# Convert to DataFrame for easier exploration
df_tfidf = pd.DataFrame(tfidf_matrix.toarray(), columns=feature_names)

print("TF-IDF Matrix Shape:", tfidf_matrix.shape)
df_tfidf.head()
```

```python
def get_top_keywords(row, feature_names, top_n=3):
    """Get top n keywords for a specific document row."""
    top_indices = row.argsort()[-top_n:][::-1]
    top_keywords = [(feature_names[i], row[i]) for i in top_indices if row[i] > 0]
    return top_keywords

print("Top Keywords for each document with their TF-IDF scores:\n")
for i, doc in enumerate(documents):
    keywords = get_top_keywords(tfidf_matrix.toarray()[i], feature_names)
    print(f"Doc {i+1}: {doc}")
    for word, score in keywords:
        print(f"  - {word}: {score:.4f}")
    print("-" * 30)
```

## Explanation of Top Keywords

TF-IDF identifies keywords that are frequent in a specific document but relatively rare across the entire collection. 

1. **Doc 2 (AI/Tech)**: Keywords like 'automation', 'intelligence', and 'transforming' stand out because they define the core subject of that specific sentence, whereas common words like 'the' are ignored (stop-words) or given low weight.
2. **Doc 4 (Space)**: 'rockets', 'private', and 'milestones' are highly specific to this document, making them the most "important" words here.
3. **Doc 5 (Climate)**: 'factors', 'factors', and 'biodiversity' highlight the specific scientific context of this document.

In short, higher scores indicate words that are more unique and representative of that particular document's content.
